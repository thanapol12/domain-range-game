### 2. อัปโหลดไฟล์ตัวเกม (เฉพาะกรณีที่รันครั้งแรกและต้องการอัปโหลด zip)

*หากอัปโหลดไฟล์เข้ามาแยกด้วยตนเองในแถบเครื่องมือซ้ายมือเสร็จแล้ว สามารถข้ามเซลล์นี้ได้เลย*

```python
from google.colab import files
import zipfile
import os

print("กรุณาอัปโหลดไฟล์ game.zip ที่ถูกบีบอัดมาจากไดเรกทอรีโครงการ:")
uploaded = files.upload()

if 'game.zip' in uploaded:
    with zipfile.ZipFile('game.zip', 'r') as zip_ref:
        zip_ref.extractall('.')
    print("✅ แตกไฟล์เกมเสร็จสิ้น!")
else:
    print("❌ ไม่พบไฟล์ game.zip กรุณาลากไฟล์เกมวางทางเมนูด้านซ้ายแทน")
```
